```python
data, sr = librosa.load("audios/voice_ads.mp3")
```

```python
data
```

```python
sr
```

```python
print(f"y: {data[:10]}")
print(f"shape y: {data.shape}")
print(f"sr: {sr}")
```

```python
pd.Series(data).plot(figsize=(10, 5),
                    lw=1,
                    title="Waveform of the audio file",
                    )
plt.show()
```

```python
plt.figure(figsize=(12, 3))
# plt.figure()
librosa.display.waveshow(data, sr=sr)
```

```python
melspec = librosa.feature.melspectrogram(y=data, sr=sr)
```

```python
log_s = librosa.amplitude_to_db(melspec)
```

```python
plt.figure(figsize=(10, 5))
librosa.display.specshow(log_s, sr=sr, x_axis='time', y_axis='log')
```
